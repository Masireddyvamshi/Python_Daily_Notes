```python
class MyClass:
    def __new__(cls,*args):
        print("in_new")
        instance=super().__new__(cls)
        return instance
    def __init__(self,list1):
        self.list1=list1
    def __add__(self,obj2):
        return [i+j for i,j in zip(self.list1,obj2.list1)]
    def __sub__(self,obj2):
        return [i-j for i,j in zip(self.list1,obj2.list1)]
    def __mul__(self, other):
        return [i * j for i, j in zip(self.list1, other.list1)]
    def __truediv__(self, other):
        return [i / j for i, j in zip(self.list1, other.list1)]
    def __floordiv__(self, other):
        return [i // j for i, j in zip(self.list1, other.list1)]
    def __mod__(self, other):
        return [i % j for i, j in zip(self.list1, other.list1)]
    def __pow__(self, other):
        return [i ** j for i, j in zip(self.list1, other.list1)]
    def __divmod__(self, other):
        return [divmod(i, j)[0] for i, j in zip(self.list1, other.list1)]
    def __contains__(self, item):
        return item in self.list1
    def __radd__(self, other):
        return [i + j for i, j in zip(other, self.list1)]
    def __rsub__(self, other):
        return [i - j for i, j in zip(other, self.list1)]
    def __rmul__(self, other):
        return [i * j for i, j in zip(other, self.list1)]
    def __rtruediv__(self, other):
        return [i / j for i, j in zip(other, self.list1)]
    def __rfloordiv__(self, other):
        return [i // j for i, j in zip(other, self.list1)]
    def __rmod__(self, other):
        return [i % j for i, j in zip(other, self.list1)]
    def __rpow__(self, other):
        return [i ** j for i, j in zip(other, self.list1)]
    def __rdivmod__(self, other):
        return [divmod(i, j)[0] for i, j in zip(other, self.list1)]
    def __iadd__(self, other):
        self.list1 = [i + j for i, j in zip(self.list1, other.list1)]
        return self
    def __isub__(self, other):
        self.list1 = [i - j for i, j in zip(self.list1, other.list1)]
        return self
    def __imul__(self, other):
        self.list1 = [i * j for i, j in zip(self.list1, other.list1)]
        return self
    def __itruediv__(self, other):
        self.list1 = [i / j for i, j in zip(self.list1, other.list1)]
        return self
    def __ifloordiv__(self, other):
        self.list1 = [i // j for i, j in zip(self.list1, other.list1)]
        return self
    def __imod__(self, other):
        self.list1 = [i % j for i, j in zip(self.list1, other.list1)]
        return self
    def __ipow__(self, other):
        self.list1 = [i ** j for i, j in zip(self.list1, other.list1)]
        return self
    def __del__(self):
        print(f"my obj list {self.list1} is destroyed")
    def __str__(self): #Gives a readable, user-friendly string representation of an object.
        return f'My class with {self.list1}'
    def __repr__(self): #__repr__ is a special method in Python that returns a string representation of an object.Its goal is to be unambiguous — meaning it should tell a developer exactly what the object is, ideally in a way that could recreate it.
        return 'my obj list is{self.list1}'
    def __matmul__(self, other):
        return sum(a*b for a,b in zip(self.list1, other.list1))
    def __imatmul__(self, other):
        self.dot = sum(a*b for a,b in zip(self.list1, other.list1))
        return self 
    def __neg__(self):
        return MyClass([-i for i in self.list1])
    def __pos__(self):
        return MyClass([+i for i in self.list1])
    def __abs__(self):
        return MyClass([abs(i) for i in self.list1])
    def __eq__(self, other): 
        return sum(self.list1) == sum(other.list1)
    def __ne__(self, other):
        return sum(self.list1) != sum(other.list1)
    def __lt__(self, other):
        return sum(self.list1) < sum(other.list1)
    def __le__(self, other):
        return sum(self.list1) <= sum(other.list1)
    def __gt__(self, other):
        return sum(self.list1) > sum(other.list1)
    def __ge__(self, other):
        return sum(self.list1) >= sum(other.list1)
    def __len__(self):
        return len(self.list1)
    def __getitem__(self, index):
        return self.list1[index]
    def __setitem__(self, index, value): 
        self.list1[index] = value
    def __delitem__(self, index): 
        del self.list1[index]
    def __iter__(self):
        self._index = 0
        return self
    def __next__(self):
        if self._index >= len(self.list1):
            raise StopIteration
        value = self.list1[self._index]
        self._index += 1
        return value
    def __call__(self):
        """Make the object callable to return the sum of its list"""
        return sum(self.list1)

# Create objects
l1 = MyClass([1, 2, 3, 4, 5])
l2 = MyClass([6, 7, 8, 9, 10])

# --- Arithmetic ---
print("Addition:", l1 + l2)
print("Subtraction:", l2 - l1)
print("Multiplication:", l1 * l2)
print("True Division:", l2 / l1)
print("Floor Division:", l2 // l1)
print("Modulo:", l2 % l1)
print("Power:", l1 ** MyClass([1,1,1,1,1]))
print("Divmod:", divmod(l2, l1))

# --- In-place Arithmetic ---
l1 += l2
print("After l1 += l2:", l1.list1)
l1 -= MyClass([1,1,1,1,1])
print("After l1 -= [1,1,1,1,1]:", l1.list1)
l1 *= MyClass([1,2,1,2,1])
print("After l1 *= [1,2,1,2,1]:", l1.list1)
l1 /= MyClass([1,1,1,1,1])
print("After l1 /= [1,1,1,1,1]:", l1.list1)
l1 //= MyClass([1,1,1,1,1])
print("After l1 //= [1,1,1,1,1]:", l1.list1)
l1 %= MyClass([2,2,2,2,2])
print("After l1 %= [2,2,2,2,2]:", l1.list1)

# --- Matmul ---
print("Matmul l1 @ l2:", l1 @ l2)
l1 @= MyClass([1,2,3,4,5])
print("After l1 @= [1,2,3,4,5]:", l1.list1)

# --- Unary ---
l3 = MyClass([1, -2, 3])
print("Unary neg:", -l3)
print("Unary pos:", +l3)
print("Absolute:", abs(l3))

# --- Comparison ---
print("Equal:", l1 == l2)
print("Not Equal:", l1 != l2)
print("Less than:", l1 < l2)
print("Less or Equal:", l1 <= l2)
print("Greater than:", l1 > l2)
print("Greater or Equal:", l1 >= l2)

# --- Container ---
print("Length l1:", len(l1))
print("l1[0]:", l1[0])
l1[0] = 99
print("After setitem l1[0] = 99:", l1.list1)
del l1[1]
print("After delitem l1[1]:", l1.list1)
print("Is 99 in l1?", 99 in l1)
print("Iterating over l1:")
for val in l1:
    print(val)

# --- String representations ---
print("str(l1):", str(l1))
print("repr(l2):", repr(l2))

print(l1.list1)
print(l1())
```

```python
class MyClass:
    def __init__(self, value):
        self.value = value
    
    def __format__(self, format_spec):
        if format_spec == "hex":
            return hex(self.value)
        elif format_spec == "bin":
            return bin(self.value)
        else:
            return str(self.value)

obj = MyClass(255)
print(f"{obj}")          # default -> calls __format__("") -> returns "255"
print(f"{obj:hex}")      # custom -> returns "0xff"
print(f"{obj:bin}")      # custom -> returns "0b11111111"
```

```python
class AttributeDemo:
    def __init__(self):
        self.existing_attr = "I exist"

    def __getattr__(self, name):
        # Called only when the attribute is NOT found normally
        print(f"__getattr__ triggered for missing attribute '{name}'")
        return None

    def __setattr__(self, name, value):
        # Called whenever an attribute is set
        print(f"__setattr__ triggered: setting '{name}' = {value}")
        super().__setattr__(name, value)

    def __delattr__(self, name):
        # Called whenever an attribute is deleted
        print(f"__delattr__ triggered: deleting '{name}'")
        super().__delattr__(name)

# Example usage
obj = AttributeDemo()

# Setting attributes
obj.new_attr = 42
obj.existing_attr = "Updated"

# Accessing attributes
print("existing_attr:", obj.existing_attr)   # Normal access
print("missing_attr:", obj.missing_attr)     # Triggers __getattr__

# Deleting attributes
del obj.new_attr
# Trying to access after deletion triggers __getattr__
print("new_attr after deletion:", obj.new_attr)
```
