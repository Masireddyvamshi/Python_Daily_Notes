# For-Else Loop Problems

### Problem 1: Write a program that iterates through a list of integers. Print "Found a negative number" and break the loop if a negative number is found. If the loop completes without finding any negative numbers,print "All numbers are non-negative."


```python
list_a=list(map(int,input().split()))
for i in list_a:
    if i<0:
        print("Found a negative number")
        break
else:
    print("All numbers are non-negative")
```

### Problem 2: Write a program to search for a specific target number (e.g., 42) in a list of integers. If the number is found, print "Target found!" and stop searching. If the loop finishes without finding the target, print "Target not found."


```python
list_a=list(map(int,input().split()))
for i in list_a:
    if i==42:
        print("Target Found")
        break
else:
    print("Target Not Found")
```

### Problem 3: Iterate through a list of employee IDs. If you find an ID that's not a multiple of 10, print "Invalid ID found" and exit the loop. If all IDs are valid (multiples of 10), print "All employee IDs are valid."


```python
employee_ids=list(map(int,input().split()))
for i in employee_ids:
    if i%10!=0:
        print("Invalid ID found")
        break
else:
    print("All employee Ids are valid")
```

### Problem 4: Given a list of tuples, where each tuple contains a name and a number (e.g., [('A', 1), ('B', 2)]), iterate through the list. If you find a number that is greater than 100, print "Found an out-of-range value" and stop. If all numbers are 100 or less, print "All values are within range."


```python
list_tuples=[('A',1),("B",20),("C",100),("D",102)]
for i in list_tuples:
    if i[1]>100:
        print("Found an out-of-range value")
        break
else:
    print("All values are within range")
```

### Problem 5: Write a program to check if all numbers in a list are perfect squares. If a non-perfect square is found, print "Not all numbers are perfect squares" and break. If all numbers are perfect squares, print "All numbers are perfect squares." You can assume a pre-existing is_perfect_square(n) function.


```python
list_perfect_notperfect=list(map(int,input().split()))
for i in list_perfect_notperfect:
    if int(i**0.5)**2!=i:
        print("Not all numbers are perfect squares")
        break
else:
    print("All numbers are perfect squares")
```

### Problem 6: Iterate through a list of lists of numbers. If an inner list contains only positive numbers, print the inner list. If the main loop completes without finding any inner list that contains only positive numbers, print "No sublist contains only positive numbers."


```python
nums=[[1,-2],[3,-4],[5,-6]]
for i in nums:
    for j  in i:
        if j<0:
            break
    else:
        print(i)
        break
else:
    print("No sublist contains only positive numbers")
            
```

### Problem 7: Given a list of integers, find the first prime number. Print the number and break the loop. If no prime number is found after checking all integers, print "No prime numbers found in the list."

```python
nums=list(map(int,input().split()))
for i in nums:
    if i<2:
        continue
    for j in range(2,(i//2)+1):
        if i%j==0 :
            break 
    else:
        print(i)
        break
else:
    print("No Prime numbers found in the list")
```

### Problem 8: Write a program that checks if a list of numbers is sorted in ascending order. If you find any number that is less than the previous one, print "List is not sorted" and stop. If the loop completes, print "List is sorted."


```python
nums=list(map(int,input().split()))
for i in range(len(nums)-1):
    if nums[i]>nums[i+1]:
        print("List is not sorted")
        break
else:
    print("List is sorted")
```

### Problem 9: Iterate through a list of numbers. For each number, check if it is divisible by both 3 and 7. If you find such a number, print "Found a number divisible by 21" and exit. If the loop finishes without finding one, print "No number is divisible by 21."


```python
nums=list(map(int,input().split()))
for i in nums:
    if i%3==0 and i%7==0:
        print("Found a number divisible by 21")
        break
else:
    print("No number is divisible by 21")
        
```

### Problem 11: Write a program to find if a list contains any duplicate numbers. If a duplicate is found, print "Duplicate found" and break. If the loop completes, print "No duplicates found."


```python
list_a=list(map(int,input().split()))
for i in list_a:
    c=list_a.count(i)
    if c>1:
        print("Duplicate found")
        break
else:
    print("No duplicates found")
```

### Problem 12: Iterate through a list of lists of integers. Find the first sublist where the sum of its elements is greater than 100. Print the sum and break. If no such sublist is found, print "No sublist has a sum greater than 100."


```python
nums=[[50,40],[1,2],[45,35]]
for i in nums:
    sum_=i[0]+i[1]
    if sum_>100:
        print(sum_)
        break
else:
    print("No sublist has a sum greater than 100")
```

### Problem 13: Write a program that checks if all numbers in a list are unique. If a number repeats, print "List contains duplicates" and break. If the loop finishes, print "All numbers are unique."


```python
nums=list(map(int,input().split()))
new_list=[]
for i in nums:
    if i in new_list:
        print("List contains duplicates")
        break
    new_list.append(i)
else:
    print("All numbers are unique")
```

### Problem 14: Iterate through a list of numbers. If you encounter a number that is both a perfect square and a perfect cube, print "Found a special number" and break. If the loop completes, print "No special numbers found."


```python
nums=list(map(int,input().split()))
for i in nums:
    if int(i**0.5)**2==i and int(i**(1/3))**3==i:
        print("Found a special number")
        break
else:
    print("No special numbers found")
```

### Problem 15: Write a program that checks if a list of numbers is a Fibonacci sequence. If you find a number that violates the Fibonacci pattern, print "Not a valid Fibonacci 

```python
nums=list(map(int,input().split()))
#length of num should be greater than 3
for i in range(2,len(nums)):
    if nums[i]!=nums[i-1]+nums[i-2]:
        print("Not a valid Fibonacci")
        break
else:
    print("valid Fibonacci Numbers")
```

### Problem 16: You have a list of lists, where each inner list contains three numbers representing the sides of a triangle. Check if all triangles are valid (sum of any two sides is greater than the third). If you find an invalid triangle, print "Invalid triangle found" and break. If all are valid, print "All triangles are valid."

```python
triangles = [[3,4,5], [1,2,4], [5,12,13]]
for a,b,c in triangles:
    if a+b<c:
        print("Invalid Triangle Found")
        break
else:
    print("All traiangles are valid")
        
```

### Problem 18: Write a program to check if any number in a list is a power of 2. If a power of 2 is found, print "Found a power of 2" and break. If the loop finishes, print "No powers of 2 found."

### Problem 19: You have a list of numbers. Check if the numbers are increasing. If you find any number that is less than the number that came before it, print "Numbers are not strictly increasing" and stop. If the loop completes, print "Numbers are strictly increasing."


```python
nums=list(map(int,input().split()))
for i in range(len(nums)-1):
    if nums[i]>nums[i+1]:
        print("Numbers are not strictly increasing")
        break
else:
    print("Numbers are strictly increasing")
```
