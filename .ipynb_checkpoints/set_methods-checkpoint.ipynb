```python
import random
class ListSetMethods:
    def convert_to_int(list_1):
        for i in range(len(list_1)):
            list_1[i]=int(list_1[i])
        return list_1
    def __init__(self,list_1 :list,list_2 :list):
        assert type(list_1)==type(list_2)==list,'Pass Only List Container'
        self.list_1=ListSetMethods.convert_to_int(list_1)
        self.list_2=ListSetMethods.convert_to_int(list_2)
    def union(self):
        merged_list=self.list_1+self.list_2
        union_list=[]
        for i in merged_list:
            if i not in union_list:
                union_list.append(i)
        return union_list
    def intersection(self):
        intersection_list=[] 
        for i in self.list_1:
            if i in self.list_2 and i not in intersection_list:
                intersection_list.append(i)
        return intersection_list
    def difference(self):
        difference_list=[]
        for i in self.list_1:
            if i not in self.list_2 and i not in difference_list:
                difference_list.append(i)
        return difference_list
    def symmetric_difference(self):
        symmetric_difference_list=[]
        for i in self.list_1:
            if i not in self.list_2 and i not in symmetric_difference_list:
                symmetric_difference_list.append(i)
        for i in self.list_2:
            if i not in self.list_1 and i not in symmetric_difference_list:
                symmetric_difference_list.append(i)
        return symmetric_difference_list
    def pop(self,list_choice):
        if list_choice==1:
            random_value=random.choice(self.list_1)
            self.list_1.remove(random_value)
        else:
            random_value=random.choice(self.list_2)
            self.list_2.remove(random_value)
    def add(self,list_choice,value):
        if list_choice==1:
            index=random.randint(0,len(self.list_1))
            self.list_1.insert(index,value)
        else:
            index=random.randint(0,len(self.list_2))
            self.list_2.insert(index,value)
    def update(self,list_choice,items):
        if list_choice==1:
            for i in items:
                index=random.randint(0,len(self.list_1))
                self.list_1.insert(index,i)
        if list_choice==2:
            for i in items:
                index=random.randint(0,len(self.list_2))
                self.list_2.insert(index,i)
        
list_1=input().split(",")
list_2=input().split(",")
List_obj1=ListSetMethods(list_1,list_2)
List_obj1.update(1,[55,99,66,55])
print(List_obj1.list_1)
# print(List_obj1.intersection())
# print(List_obj1.difference())
# print(List_obj1.symmetric_difference())
```
