### 1.You are given two lists, fruits = ['apple', 'banana', 'cherry'] and prices = [1.20, 0.50, 2.75].

```python
fruits = ['apple', 'banana', 'cherry'] 
prices = [1.20, 0.50, 2.75]
fruit_prices=["The price of a(n) {} is ${}.".format(i,j) for i,j in zip(fruits,prices)]
for i in fruit_prices:
    print(i)
    
```

### Q1.Generate a list where numbers from 1 to 15 are replaced with "Even" if even, otherwise "Odd".


```python
a=int(input())
b=int(input())
a,b=min(a,b),max(a,b)
list_a=["Even" if i%2==0 else "Odd" for i in range(a,b+1)]
print(list_a)
```

###  Q2.From a list [12, 7, 19, 28, 33], create a new list with "Div3" if divisible by 3, otherwise the number itself.

```python
list_a=[12,7,19,28,33]
new_list=["Div3" if i%3==0 else i for i in list_a]
print(new_list)
```

### Q24.For numbers 1–30, return: "Div2", "Div3", "Div5", "Div2&3", "Div2&5", "Div3&5", "Div2&3&5", else "None".

###  Q25.From scores = [95, 82, 76, 54, 43, 22], grade: A (≥90), B (80–89), C (70–79), D (50–69), E (40–49), F (<40). Then add "Pass" or "Fail" based on cutoff ≥50.


```python
scores = [95, 82, 76, 54, 43, 22]
grades_and_results=[("A" if s >= 90 else "B" if s >= 80 else "C" if s >= 70 else "D" if s >= 50 else "E" if s >= 40 else "F") + ("-Pass" if s >= 50 else "-Fail") for s in scores ]
print(grades_and_results)
```
