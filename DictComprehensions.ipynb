# Dictionary Comprehension Problems


### Problem 1: Given a list of integers, create a dictionary where the keys are the original numbers. The values should be "Even" if the number is even and positive, "Odd" if it's odd and positive, and "Negative" if it's negative.


```python
list_a=list(map(int,input().split(",")))
dict_a={i:"Even" if i>0 and i%2==0 else "Odd" if i>0 else "Negative"   for i in list_a}
print(dict_a)
```

### Problem 2: Given a list of integers, create a dictionary where each number is a key. The value should be the number itself if it's a multiple of 5, the number squared if it's a multiple of 3, and the number cubed if it's neither.


```python
list_a=list(map(int,input().split()))
dict_a={i:i if i%5==0 else i**2 if i%3==0 else i*3 for i in list_a}
print(dict_a)
```

### Problem 3: Given a list of integers, create a dictionary. For each number, the key should be the number and the value should be True if it's a prime number, False otherwise. You can assume you have a pre-existing function is_prime(n) to use.


```python
list_a=list(map(int,input().split()))
dict_a={i:False if i<=1 else True if  [j for j in range(2,i) if i%j==0]==[] else "False"  for i in list_a}
print(dict_a)
```

### Problem 4: Given a list of temperatures in Celsius, create a dictionary where the keys are the original temperatures. The values should be "Hot" if the temperature is above 30, "Warm" if between 15 and 30 (inclusive), and "Cold" otherwise.

```python
list_of_temperatures=list(map(int,input().split())) 
dict_temperatures={i:"Hot" if i>30 else "Warm" if 15<=i<=30 else "Cold"  for i in list_of_temperatures}
print(dict_temperatures)
```

### Problem 5: Given a list of integers, create a dictionary where the keys are the numbers. The values should be the number's square root if it's a perfect square, its half if it's an even number, and its cube otherwise.


```python
list_a=list(map(int,input().split()))
dict_a={n:n**0.5 if  (n**0.5)**2==n else n//2 if n%2==0 else n**3  for n in list_a}
print(dict_a)
```

### Problem 6: Given a list of integers, create a dictionary where the keys are the numbers. The values should be 1 if the number is in the range 1-10, 2 if in the range 11-20, 3 if in the range 21-30, and 0 otherwise.

```python
list_a=list(map(int,input().split()))
dict_a={i:1 if i in range(1,11) else 2 if i in range(11,21) else 3 if i in range(21,31) else 3 for i in list_a}
print(dict_a)
```

### Problem 7: Given a list of integers, create a dictionary. The keys should be the numbers, and the values should be "Fizz" if the number is a multiple of 3, "Buzz" if it's a multiple of 5, and "FizzBuzz" if it's a multiple of both. If it's none of these, the value should be the number itself.


```python
list_a=list(map(int,input().split()))
dict_multiples={i:"FizzBuzz" if i%3==0 and i%5==0 
                else "Fizz" if i%3==0 
                else "Buzz" if i%5==0 
                else None 
                for i in list_a}
print(dict_multiples)
```

### Problem 8: Given a list of numbers, create a dictionary. The keys should be the numbers, and the values should be "Positive Even" if the number is positive and even, "Negative Even" if negative and even, "Positive Odd" if positive and odd, and "Negative Odd" if negative and odd.


```python
list_a=list(map(int,input().split()))
dict_even_odd={i:"Postive Even" if i%2==0 and i>0
               else "Postive Odd" if i>0
               else "Negative Even" if i%2==0 and i<0
               else "Negative Odd"
               for i in list_a}
print(dict_even_odd)
```

### Problem 9: Given a list of lists of integers, where each inner list has two elements (e.g., [[1, 2], [3, 4]]), create a dictionary. The key should be the first element of the inner list, and the value should be True if the second element is greater than the first, False otherwise.


```python
list_a=[[1,2],[3,4],[6,5]]
dict_of_inner_list={i[0]:True if i[1]>i[0] 
                    else False
                    for i in list_a}
print(dict_of_inner_list)
```

### Problem 10: Given a list of integers, create a dictionary where the keys are the original numbers. The values should be "High Odd" if the number is greater than 100 and odd, "Low Odd" if it's less than 100 and odd, and "Even" if it's even.


```python
list_a=list(map(int,input().split()))
dict_original_numbers={i:"High Odd" if i>100 and i%2!=0 
                       else "Low Odd" if i<100 and i%2!=0
                       else "Even" 
                       for i in list_a}
print(dict_original_numbers)
```

### Problem 11: Given a list of integer scores, create a dictionary where the keys are the scores. The values should be "A" if the score is over 90, "B" if over 80, "C" if over 70, and "F" otherwise.

### Problem 12: Given a list of numbers, create a dictionary where the keys are the numbers. The values should be "Divisible by 4" if the number is a multiple of 4, "Divisible by 2" if it's a multiple of 2 but not 4, and "Neither" otherwise.


```python
list_a=list(map(int,input().split())) 
dict_divisiblenums={i:"Divisible by 4" if i%4==0
                    else "Divisible by 2" if i%2==0
                    else "Neither"
                    for i in list_a}
print(dict_divisiblenums)
```

### Problem 13: Given a list of tuples, where each tuple contains two integers (e.g., [(5, 2), (8, 9)]), create a dictionary. The key should be the sum of the two integers, and the value should be the product of the two integers if the sum is even, and the difference if the sum is odd.


```python
list_of_tuples=[(5,2),(8,9),(2,4)]
dict_a={i[0]+i[1]:i[0]*i[1] if (i[0]+i[1])%2==0
        else i[0]-i[1]
        for i in list_of_tuples}
print(dict_a)
```

### Problem 14: Iterate through a list of numbers. If you encounter a number that is both a perfect square and a perfect cube, print "Found a special number" and break. If the loop completes, print "No special numbers found."


```python
list_a=list(map(int,input().split())) 
dict_a=[i for i in list_a if (i**0.5)**2==i and (i**(1/3))**3==i]
if dict_a:
    print("Found a special number")
else:
    print("No special numbers found")
```

### Problem 15: Given a list of numbers, create a dictionary where the keys are the numbers. The values should be "Positive" if the number is positive and a multiple of 10, "Negative" if negative and a multiple of 10, and "Zero or Not a Multiple of 10" otherwise.

### Problem 16: Given a list of integers, create a dictionary. For each number, the key is the number, and the value is its half if the number is a perfect square, and its double if it's a multiple of 3, otherwise the number itself.


```python
list_a=list(map(int,input().split()))
dict_key_value_pair={i:(i/2) if int(i**0.5)**2==i
                     else i+i if i%3==0
                     else i 
                     for i in list_a}
print(dict_key_value_pair)
```

### Problem 17: Given a list of lists of numbers (e.g., [[10, 20], [5, 15]]), create a dictionary. The key for each item is the sum of the inner list, and the value is True if the first element is larger than the second, and False otherwise.


```python
a=[[10, 20], [5, 15],[30,20]]
dict_a={i[0]+i[1]: True if i[0]>i[1] 
        else False
        for i in a}
print(dict_a)
```

### Problem 18: Given a list of integers, create a dictionary. The keys are the numbers. The values should be 1 if the number is a power of 2, 2 if a power of 3, and 0 otherwise. You can assume a pre-existing is_power(base, n) function.


```python
list_a=list(map(int,input().split()))
dict_a={i:1 if i%2==0 and i>0 
        else 2 if i%3==0 and i>0 
        else 0 
        for i in list_a}
print(dict_a)
```

### Problem 19: Given a list of integers, create a dictionary where keys are the numbers. The values should be "Valid" if the number is between 1 and 100, inclusive. Otherwise, if it is less than 1, the value is "Too Small", and if it is greater than 100, the value is "Too Large".

```python
list_a=list(map(int,input().split()))
dict_a={i:"Valid" if 1<=i<=100 
        else "Too Small" if i<1
        else "Too Large"
        for  i in list_a}
print(dict_a)
```
