# Addition operator


```python
x=2
print(x)
print(id(x))
x+=4
print(x)
print(id(x))
```

# Subraction operator

```python
x=2
print(x)
print(id(x))
x-=2
print(x)
print(id(x))
```

# division operator

```python
x=4
print(x)
print(id(x))
x/=2
print(x)
print(id(x))
```

# Floor division operator

```python
x=4
print(x)
print(id(x))
x//=2
print(x)
print(id(x))
```

# Remainder

```python
x=4
print(x)
print(id(x))
x%=2
print(x)
print(id(x))
```

# Exponent

```python
x=4
print(x)
print(id(x))
x**=2
print(x)
print(id(x))
```

# Multiplication

```python
x=4
print(x)
print(id(x))
x*=2
print(x)
print(id(x))
```
