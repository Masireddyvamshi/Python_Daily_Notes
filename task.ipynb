```python
import copy

class MyCollection(dict):   # Inherit from dict to use __missing__
    def __init__(self, items):
        super().__init__(items)
        self.data = list(items.values())  # store values as list
    
    # 1. Hashing (so objects can be dict keys or in sets)
    def __hash__(self):
        return hash(tuple(self.data))
    
    # 2. Shallow copy
    def __copy__(self):
        print("Shallow copy called")
        return MyCollection(self.copy())
    
    # 3. Deep copy
    def __deepcopy__(self, memo):
        print("Deep copy called")
        return MyCollection(copy.deepcopy(dict(self), memo))
    
    # 4. Reversed iteration
    def __reversed__(self):
        return iter(self.data[::-1])
    
    # 5. Missing key handling
    def __missing__(self, key):
        return f"{key} is missing!"

# ------------------ Testing ------------------
obj = MyCollection({'a': 1, 'b': 2, 'c': 3})

# hash()
print("Hash:", hash(obj))

# copy
shallow = copy.copy(obj)   # triggers __copy__
deep = copy.deepcopy(obj)  # triggers __deepcopy__

# reversed
print("Reversed:", list(reversed(obj)))

# missing key
print("Missing key:", obj['z'])
```

```python
class MyClass:
    def __init__(self, value):
        self.value = value
    
    def __format__(self, format_spec):
        if format_spec == "hex":
            return hex(self.value)
        elif format_spec == "bin":
            return bin(self.value)
        else:
            return str(self.value)

obj = MyClass(255)
print(f"{obj}")          # default -> calls __format__("") -> returns "255"
print(f"{obj:hex}")      # custom -> returns "0xff"
print(f"{obj:bin}")      # custom -> returns "0b11111111"
```

```python
class AttributeDemo:
    def __init__(self):
        self.existing_attr = "I exist"

    def __getattr__(self, name):
        # Called only when the attribute is NOT found normally
        print(f"__getattr__ triggered for missing attribute '{name}'")
        return None

    def __setattr__(self, name, value):
        # Called whenever an attribute is set
        print(f"__setattr__ triggered: setting '{name}' = {value}")
        super().__setattr__(name, value)

    def __delattr__(self, name):
        # Called whenever an attribute is deleted
        print(f"__delattr__ triggered: deleting '{name}'")
        super().__delattr__(name)

# Example usage
obj = AttributeDemo()

# Setting attributes
obj.new_attr = 42
obj.existing_attr = "Updated"

# Accessing attributes
print("existing_attr:", obj.existing_attr)   # Normal access
print("missing_attr:", obj.missing_attr)     # Triggers __getattr__

# Deleting attributes
del obj.new_attr
# Trying to access after deletion triggers __getattr__
print("new_attr after deletion:", obj.new_attr)
```
