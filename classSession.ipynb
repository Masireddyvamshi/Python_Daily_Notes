# partition
Splits the string into 3 parts:

The part before the separator

The separator itself

The part after the separator

Always returns a tuple of 3 strings.

If the separator is not found → returns (original_string, '', '').

```python
'malyalam'.partition('y')
```

```python
''.join('malyalam'.partition('y'))
```

```python
'malyalam'.partition('zz')
```

# formating


```python
from string import Template
name='vamshi'
age=22
pi=3.1456
print("i am %s and i am  %d years old" %(name,age))
print("i am {0} and i am {1} years old".format(name,age))
print("i am {n} and i am {a} years old".format(n=name,a=age))
print("i am {} and i am {} years old".format(name,age))
print(f"i am {name} and i am {age} years old")
print(f"pi value is {pi:.2f}")
t=Template("i am $name and i am $age years old")
print(t.substitute(name=name,age=age))
```

```python
def dictionary(str):
    if len(str)%2!=0:
        return "pass the string where the length of str is even"
    dict_a={}
    for i in range(len(str)-1):
        if str[i] not in dict_a:
            dict_a[str[i]]=''
        dict_a[str[i]]=str[i+1]
    return dict_a
a='abacbd'
r=dictionary(a)
print(r)
```

```python
print("123".isnumeric())   
print("²".isnumeric())    
print("⑦".isnumeric())     
print("½".isnumeric())     
print("Ⅻ".isnumeric())    
print("12a3".isnumeric())  
```

```python
print("123".isdecimal())   # True
print("²".isdecimal())     # False (superscript 2 is not a decimal digit)
print("⑦".isdecimal())     # False (circled 7 is not decimal)
print("½".isdecimal())
```

```python
print("Hello123".isascii())    # True
print("123".isascii())         # True
print("π".isascii())           # False (pi is not ASCII)
print("½".isascii())           # False (fraction is not ASCII)
print("©".isascii())
```

```python
print("   ".isspace())     # True (spaces only)
print("\t\n".isspace())    # True (tab + newline)
print(" abc ".isspace())   # False (contains letters)
print("".isspace())        # False (empty string → no characters at all)
```

```python
print("hello".isidentifier())    # True
print("hello_world".isidentifier())  # True
print("123abc".isidentifier())   # False (starts with digit)
print("my var".isidentifier())   # False (space not allowed)
```

```python
print("Hello 123!".isprintable())   # True
print("Hello\nWorld".isprintable()) # False (newline is not printable)
print(" ".isprintable())            # True (space is printable)
```

```python
print("Hello World".istitle())   # True
print("Hello world".istitle())   # False ("world" not capitalized)
print("HELLO WORLD".istitle())   # False (all uppercase)
print("Hello-World".istitle())   # True (hyphen is treated as separator)
```

```python
from multipledispatch import dispatch
```

```python
class Details:
    @dispatch(str,int)
    def __init__(self,fullName,age):
        self.fullName=fullName
        self.age=age
        self.email=fullName+'@gmail.com'
    @dispatch(str,str,int)
    def __init__(self,fname,lname,age):
        self.fname=fname
        self.lname=lname
        self.age=age
        self.email=fname+lname+'@gmail.com'
a=Details('vamshidhar','Reddy',22)
a.email
```

```python
class Myclass:
    def __init__(self,data):
        self.data=data
    def __add__(self,other):
        convTuple=list(other.data)
        return self.data+convTuple
a=Myclass([1,2,3,3])
b=Myclass((1,2,3,4))
print(a+b)
```

```python
class Details:
    def __init__(self,fname,lname,age):
        self.fname=fname
        self.lname=lname
        self.age=age
        self.emails=fname+lname+'@gmail.com'
    @property
    def email(self):
        return self.emails
    @email.setter
    def email(self,new_email):
        self.emails=new_email
    @email.deleter
    def email(self):
        del self.emails
a=Details('vamshidhar','Reddy',22)
a.email
a.email='vamshidhar@gmail.com'
a.email
del a.email
a.email
```
