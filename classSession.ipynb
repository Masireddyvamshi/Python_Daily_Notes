```python
def flatten_list(lst):
    # Base case: empty list
    if not lst:
        return []
    
    first, rest = lst[0], lst[1:]

    # If first element is a list → recurse into it
    if isinstance(first, list):
        return flatten_list(first) + flatten_list(rest)
    else:
        return [first] + flatten_list(rest)


# Example usage
nested = [1, 2, 3, 4, [[]], [], ['a', 'b', 'c'], [[8, 9]], 'f']
print(flatten_list(nested))
```

```python
class CountDown:
    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= 0:
            raise StopIteration
        else:
            val = self.current
            self.current -= 1
            return val

counter = CountDown(3)
for num in counter:
    print(num)
```

```python
try:
    with open("data.txt", "r") as f:
        content = f.read()
        print(content)
except FileNotFoundError:
    print("File not found!")
except Exception as e:
    print("Error:", e)
```

```python
try:
    with open("data.txt", "w") as f:
        f.write("Overwritten content!\n")
    print("Write successful.")
except Exception as e:
    print("Error:", e)
```

```python
try:
    with open("data.txt", "a") as f:
        f.write("Appended line.\n")
    print("Append successful.")
except Exception as e:
    print("Error:", e)
```

```python
try:
    with open("newfile.txt", "x") as f:
        f.write("Created new file.")
    print("File created.")
except FileExistsError:
    print("File already exists!")
```

```python
try:
    with open("image.png", "rb") as f:
        data = f.read()
    print("Binary data length:", len(data))
except Exception as e:
    print("Error:", e)
```
